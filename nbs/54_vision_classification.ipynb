## Setup

Import dependencies for vision examples.

```python
import warnings

warnings.filterwarnings('ignore')

import pandas as pd
from datasets import load_dataset

from blurr.vision.classification import VisionClassificationLearner
```

## Quick Start

Train a compact image classifier from a Hugging Face image dataset.

```python
learn = VisionClassificationLearner.from_hf(
    'beans',
    model_name='google/vit-base-patch16-224',
    image_col='image',
    label_col='labels',
    max_samples=200,
    bs=8,
)

learn.fit_one_cycle(1, 2e-4)

beans = load_dataset('beans')
sample_image = beans['validation'][0]['image']
quick_pred = learn.blurr_predict(sample_image)[0]
quick_pred
```

**Expected output**

A dictionary with `label`, `score`, `probs`, and `class_labels` for the input image.

## API Reference

### Main class
- `VisionClassificationLearner`

### Construction methods
- `from_hf(dataset, model_name, ...)`
- `from_df(df, model_name, ...)`
- `from_data(dls, model_name, ...)` for prebuilt DataLoaders
- `from_data_source(data, model_name, ...)` for DataFrame/HF dispatch

### Inference
- `blurr_predict(images, threshold=0.5)`
